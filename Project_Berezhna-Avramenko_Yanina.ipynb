```python
#завдання 1
total_respondents = df["ResponseId"].nunique()

print(total_respondents)
```

```python
#завдання 2
questions = set(df1["qname"].unique())
questions = questions.intersection(df.columns)
complete_responses = df[list(questions)].dropna().shape[0]

print(complete_responses)
```

```python
#завдання 3
print('Міри центральної тенденції:')
# Середнє
print(f'Mean: {round(df["WorkExp"].mean(), 1)}')
# Медіана
print(f'Median: {df["WorkExp"].median()}')
# Мода
print(f'Mode: {df["WorkExp"].mode()[0]}')
```

```python
#завдання 4
remote_workers = df[df["RemoteWork"] == "Remote"].shape[0]

print(remote_workers)
```

```python
#завдання 5
worked_with_python = df["LanguageHaveWorkedWith"].str.contains(
    "Python",
    case=False,
    na=False
)

python_percentage = (worked_with_python.sum() / len(df)) * 100

print(f"Відсоток респондентів: {round(python_percentage, 1)}%")
```

```python
#завдання 6
learned_with_online_courses = df["LearnCode"].str.contains(
    "Online Courses",
    case=False,
    na=False
)

online_courses_count = learned_with_online_courses.sum()

print(online_courses_count)
```

```python
#завдання 7
worked_with_python = df[
    df["LanguageHaveWorkedWith"].str.contains(
        "Python",
        case=False,
        na=False
    )
]
worked_with_python = worked_with_python.dropna(
    subset=["ConvertedCompYearly"]
)
py_progerrs_by_country = worked_with_python.groupby("Country")[
    "ConvertedCompYearly"
].agg(["mean", "median"])

print(py_progerrs_by_country)
```

```python
#завдання 8
top_paid = df[
    ["ResponseId", "EdLevel", "ConvertedCompYearly"]
    ]
top_paid = top_paid.dropna(subset=["ConvertedCompYearly"])
top_paid = top_paid.sort_values(
    by="ConvertedCompYearly",
    ascending=False
)
top_5 = top_paid.head(5).reset_index(drop=True)
print(top_5)
```

```python
#завдання 9
df["UsesPython"] = df["LanguageHaveWorkedWith"].str.contains(
    "Python",
    case=False,
    na=False
)

python_by_age = df.groupby("Age").agg(
    total_respondents=("ResponseId", "count"),
    python_users=("UsesPython", "sum")
)

python_by_age["python_percentage"] = (
    python_by_age["python_users"]
    / python_by_age["total_respondents"]
) * 100

python_by_age["python_percentage"] = (
    python_by_age["python_percentage"].round(1)
)

print(python_by_age[["python_percentage"]])
```

```python
#завдання 9
salary_75 = df["ConvertedCompYearly"].quantile(0.75)
high_paid_remote = df[
    (df["ConvertedCompYearly"] > salary_75)
    &
    (df["RemoteWork"] == "Remote")
]
top_industries = high_paid_remote["Industry"].value_counts()

print(top_industries)
```
